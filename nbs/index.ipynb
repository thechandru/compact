## Goal

Lisp is well-suited for symbolic computation. Since it represents code and data identically (homoiconicity), it makes it trivial to embed problem-specific semantics and language inside a Lisp program. Lisp programs manipulate (evaluate and apply) lists. Since Lisp programs are themselves lists, they can "rewrite themselves". Lisp macros are functions that transform code at load time. This enables addition of new syntax that doesn't exist in the language. For example, a programmer could invent an special looping construct, conditionals (like try/catch) etc. 

One barrier to wider adoption of Lisp is limited library availability (for instance compared to Python). 

`compact` sidesteps the library problem by embedding Lisp inside Python via a shared namespace. `"(expr)" @ lisp` evaluates a Lisp expression against the calling Python frame's globals — Python variables are visible in Lisp, and anything `define`d in Lisp lands back in Python as a regular variable. Python functions are callable from Lisp directly; Lisp functions are callable from Python.

### Documentation

```python
# Python → Lisp: variables defined in Python are visible in Lisp
from compact import lisp

a, b = 3, 4
"(define c (sqrt (+ (* a a) (* b b))))" @ lisp
print(c)  # 5.0
```

```python
# Lisp → Python: TCO lets Lisp recurse where Python would stack overflow
from compact import lisp

"""
(define (factorial n acc)
  (if (= n 0) acc
      (factorial (- n 1) (* n acc))))
""" @ lisp
factorial(10000, 1)  # works fine; Python would hit recursion limit
```

```python
# Both together: Python stdlib available in Lisp, Lisp logic callable from Python
from compact import lisp
import math

pi = math.pi
"(define (circle-area r) (* pi r r))" @ lisp
sorted(map(circle_area, [1, 2, 3, 4, 5]))  # [3.14, 12.56, 28.27, 50.26, 78.53]
```
