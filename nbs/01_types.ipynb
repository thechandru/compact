###### start-section

```
#| export
from fastcore.basics import patch, basic_repr
```

```
#| export
class Pair:
    "a cons cell"
    __slots__ = ("car", "cdr")
    def __init__(self, car, cdr): self.car,self.cdr = car,cdr
    def __eq__(self, other): return isinstance(other, Pair) and self.car == other.car and self.cdr == other.cdr
    __repr__ = basic_repr("car,cdr")

class NilType:
    def __str__(self): return "()"
    def __repr__(self): return str(self)

Nil = NilType()
```

```
Pair(1, 2)
```

```
#| export
cons = lambda x,y: Pair(x, y)
car = lambda x: x.car
cdr = lambda x: x.cdr
```

```
cons(1, 2)
```

```
car(cons(1, 2))
```

```
cdr(cons(1,2))
```

```
Pair(1, Pair(2, Pair(3, Nil)))
```

```
#| export
def list2pair(lst, tail=Nil):
    "convert a python list into Scheme list / pair chain, ending in optional tail"
    if not isinstance(lst, list): return lst
    return tail if not lst else Pair(list2pair(lst[0]), list2pair(lst[1:], tail))
```

```
list2pair([1, 2, 3])
```

```
list2pair([1, 2, [101, 102], 3])
```

```
#| export
def pair_str(p):
    "string representation of a pair"
    if cdr(p) is Nil: return str(car(p))
    if isinstance(cdr(p), Pair): return f"{car(p)} {pair_str(cdr(p))}"

    return f"{car(p)} . {cdr(p)}"    
```

```
pair_str(Pair(1, 2))
```

```
pair_str(list2pair([1, 2, 3]))
```

```
#| export
@patch
def __str__(self:Pair): return f"({pair_str(self)})"
```

```
str(Pair(1, 2))
```

```
str(list2pair([1, 2, 3]))
```

```
#| export
def is_list(x): return x is Nil or (isinstance(x, Pair) and is_list(cdr(x)))
def is_pair(x): return isinstance(x, Pair)

pair_primitives = {
    "list": lambda *xs: list2pair(list(xs)),
    "cons": cons, 
    "car": car, 
    "cdr": cdr, 
    "null?": lambda x: x is Nil, 
    "pair?": is_pair,
    "list?": is_list,
    }
```
